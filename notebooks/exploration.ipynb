```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

import sympy as sp
from quantum_group import (
    QuantumGroupSL2,
    q_integer, q_factorial, q_binomial, classical_limit,
    build_representation, highest_weight_vector,
    plot_weight_diagram, plot_crystal_graph, plot_combined,
    crystal_string,
)
sp.init_printing()
```

## 1. q-arithmetic

Display `[n]_q` and `[n]_q!` and their `q → 1` limits.

```python
for n in range(6):
    qn = sp.simplify(q_integer(n))
    print(f'[{n}]_q = {qn}    classical limit = {classical_limit(qn)}')

print()
print('[5]_q! =', q_factorial(5))
print('[6 choose 2]_q =', q_binomial(6, 2))
```

## 2. Defining relations

The `QuantumGroupSL2` facade class records the four relations.

```python
Uq = QuantumGroupSL2()
Uq.print_relations()
```

## 3. Constructing the V_n representation

For n = 3 (dimension 4), inspect the E, F, K and K^{-1} matrices.

```python
rep = build_representation(3)
print('dimension =', rep.dim)
print('K =')
rep.K
```

```python
rep.E
```

```python
rep.F
```

On the highest-weight vector `v_0`, E should give zero and K should have eigenvalue q^n.

```python
v0 = highest_weight_vector(rep)
print('E . v_0 =', (rep.E * v0).T)
print('K . v_0 =', (rep.K * v0).T)
```

## 4. Verifying the relations

```python
checks = Uq.verify(rep)
for name, c in checks.items():
    print(f'{name}: {c.holds}')
```

## 5. Visualization: weight diagram and crystal

The K-weights of V_n are `n - 2k`, k = 0..n. The crystal B(n) is the directed path b_0 → b_1 → ... → b_n.

```python
%matplotlib inline
fig = plot_combined(4)
```

```python
print(crystal_string(5))
```

## 6. Classical limit (q → 1)

Since K . v_k = q^{n-2k} v_k, the eigenvalues approach 1 as q → 1; however, the operator h := log(K)/log(q) gives the classical weights n - 2k.

```python
for k in range(rep.dim):
    w = rep.weights[k]
    print(f'v_{k}: K-eigenvalue = {w}, q->1 limit = {classical_limit(w)}')
```
