## 1. q-aritmetik

`[n]_q` ve `[n]_q!` formlarını ve `q → 1` limitlerini gösterelim.

```python
for n in range(6):
    qn = sp.simplify(q_integer(n))
    print(f'[{n}]_q = {qn}    classical limit = {classical_limit(qn)}')

print()
print('[5]_q! =', q_factorial(5))
print('[6 choose 2]_q =', q_binomial(6, 2))
```

## 2. Tanımlayıcı bağıntılar

`QuantumGroupSL2` cephe sınıfı, dört bağıntıyı kayıt eder.

```python
Uq = QuantumGroupSL2()
Uq.print_relations()
```

## 3. V_n temsilinin inşası

n = 3 için (boyut 4) E, F, K, K^{-1} matrisleri.

```python
rep = build_representation(3)
print('boyut =', rep.dim)
print('K =')
rep.K
```

```python
rep.E
```

```python
rep.F
```

En yüksek ağırlık vektörü `v_0` üzerinde E sıfır vermeli, K ise q^n özdeğerini vermelidir.

```python
v0 = highest_weight_vector(rep)
print('E . v_0 =', (rep.E * v0).T)
print('K . v_0 =', (rep.K * v0).T)
```

## 4. Bağıntıların doğrulanması

```python
checks = Uq.verify(rep)
for name, c in checks.items():
    print(f'{name}: {c.holds}')
```

## 5. Görselleştirme: ağırlık diyagramı ve kristal

V_n'in K-ağırlıkları `n - 2k`, k = 0..n. Kristal B(n) ise b_0 → b_1 → ... → b_n yönlü yolu.

```python
%matplotlib inline
fig = plot_combined(4)
```

```python
print(crystal_string(5))
```

## 6. Klasik limit (q → 1)

K . v_k = q^{n-2k} v_k olduğu için q → 1 alındığında özdeğerler 1'e gider; ancak h := log(K)/log(q) operatörü n - 2k klasik ağırlıklarını verir.

```python
for k in range(rep.dim):
    w = rep.weights[k]
    print(f'v_{k}: K-özdeğer = {w}, q->1 limit = {classical_limit(w)}')
```
